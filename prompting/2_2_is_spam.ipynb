```python
import pandas as pd
from tqdm import tqdm
```

```python
df = pd.read_csv('https://stepik.org/media/attachments/lesson/1110806/100_crypto_messages.csv')
```

```python
#display(df.head(20))
```

```python
from langchain_classic import PromptTemplate
```

```python
prompt_template = """Ты специалист в борьбе со спамами в чате про крипту.
Определи из контекста, является ли он спамом.


Context: {context}

Question: Это спам?

Answer: Если спам верни 1, если нет - верни 0. Ничего кроме цифр"""
```

```python
is_spam = [] # Список, где будем хранить ответы модели

for context in tqdm(df['text']):
    prompt = prompt_template.format(context=context) # Добавляем сообщение в промпт
    amount = llm.invoke(prompt).content # Ответ модели
    is_spam.append(amount) # Добавляем ответ в список
    #break # Для отладки. Уберите, когда убедитесь, что на одном примере работает
```
